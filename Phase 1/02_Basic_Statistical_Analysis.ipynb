### Objective

The objective of this notebook is to perform basic statistical analysis on the Indian Food dataset. Descriptive statistics such as mean, median, standard deviation, variance, minimum, maximum, range, quartiles, and interquartile range will be calculated for numerical variables. Frequency distributions and unique-value counts will also be examined for categorical variables.

```python
# ============================================================
# EDA PROJECT - BASIC STATISTICAL ANALYSIS
# Name        : ANINDO PRITOM BARUA
# Registration: 23BDS0328
# Dataset     : Indian Food Dataset
# ============================================================

# Import required libraries
import pandas as pd
import numpy as np

# ------------------------------------------------------------
# 1. Load the dataset
# ------------------------------------------------------------

file_path = "indian_food_23BDS0328_cleaned.csv"

df = pd.read_csv(file_path)

print("=" * 70)
print("BASIC STATISTICAL ANALYSIS")
print("=" * 70)

# ------------------------------------------------------------
# 2. Identify numerical and categorical columns
# ------------------------------------------------------------

numerical_columns = df.select_dtypes(include=np.number).columns.tolist()
categorical_columns = df.select_dtypes(include="object").columns.tolist()

print("\nNUMERICAL COLUMNS")
print("=" * 70)
print(numerical_columns)

print("\nCATEGORICAL COLUMNS")
print("=" * 70)
print(categorical_columns)

# ------------------------------------------------------------
# 3. Descriptive statistics for numerical variables
# ------------------------------------------------------------

print("\nDESCRIPTIVE STATISTICS")
print("=" * 70)

statistics = pd.DataFrame({
    "Mean": df[numerical_columns].mean(),
    "Median": df[numerical_columns].median(),
    "Standard Deviation": df[numerical_columns].std(),
    "Variance": df[numerical_columns].var(),
    "Minimum": df[numerical_columns].min(),
    "Maximum": df[numerical_columns].max(),
    "Range": df[numerical_columns].max() - df[numerical_columns].min(),
    "IQR": df[numerical_columns].quantile(0.75)
            - df[numerical_columns].quantile(0.25)
})

display(statistics)

# ------------------------------------------------------------
# 4. Quartiles and percentiles
# ------------------------------------------------------------

print("\nQUARTILES AND PERCENTILES")
print("=" * 70)

quartiles = df[numerical_columns].quantile(
    [0.25, 0.50, 0.75]
)

quartiles.index = [
    "25th Percentile (Q1)",
    "50th Percentile (Median)",
    "75th Percentile (Q3)"
]

display(quartiles)

# ------------------------------------------------------------
# 5. Mode of numerical variables
# ------------------------------------------------------------

print("\nMODE OF NUMERICAL VARIABLES")
print("=" * 70)

for column in numerical_columns:
    mode_values = df[column].mode()

    if len(mode_values) > 0:
        print(f"{column}: {mode_values.tolist()}")

# ------------------------------------------------------------
# 6. Frequency distribution of categorical variables
# ------------------------------------------------------------

print("\nFREQUENCY DISTRIBUTIONS")
print("=" * 70)

for column in categorical_columns:
    print(f"\n--- {column} ---")
    print(df[column].value_counts(dropna=False))

# ------------------------------------------------------------
# 7. Number of unique values in each column
# ------------------------------------------------------------

print("\nUNIQUE VALUE COUNTS")
print("=" * 70)

unique_counts = pd.DataFrame({
    "Column": df.columns,
    "Unique Values": [df[column].nunique(dropna=False) for column in df.columns]
})

display(unique_counts)

# ------------------------------------------------------------
# 8. Most common value of categorical variables
# ------------------------------------------------------------

print("\nMOST FREQUENT CATEGORY")
print("=" * 70)

for column in categorical_columns:
    mode_values = df[column].mode()

    if len(mode_values) > 0:
        print(f"{column}: {mode_values.iloc[0]}")

# ------------------------------------------------------------
# 9. Statistical summary using Pandas
# ------------------------------------------------------------

print("\nPANDAS DESCRIPTIVE SUMMARY")
print("=" * 70)

display(df.describe())

# ------------------------------------------------------------
# 10. Final summary
# ------------------------------------------------------------

print("\n" + "=" * 70)
print("BASIC STATISTICAL ANALYSIS COMPLETED")
print("=" * 70)
```

## Conclusion

Basic statistical analysis was successfully performed on the Indian Food dataset.

The numerical variables were analyzed using measures of central tendency and dispersion, including mean, median, standard deviation, variance, minimum, maximum, range, quartiles, and interquartile range.

The categorical variables were examined using frequency distributions, mode, and unique-value counts. These statistical measures provide an overall understanding of the distribution and characteristics of the dataset.

The results obtained from this analysis will be useful for the subsequent stages of data cleaning, transformation, and exploratory visualization.
