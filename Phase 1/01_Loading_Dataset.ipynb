### Objective

The objective of this notebook is to load the Indian Food dataset into Python and perform an initial inspection of the dataset, including its structure, dimensions, column names, data types, statistical summary, and sample records.

```python
# ============================================================
# EDA PROJECT - LOADING DATASET
# Name       : ANINDO PRITOM BARUA
# Registration: 23BDS0328
# Dataset    : Indian Food Dataset
# ============================================================

# Import required libraries
import pandas as pd
import numpy as np

# ------------------------------------------------------------
# 1. Load the dataset
# ------------------------------------------------------------

file_path = "indian_food_23BDS0328_cleaned.csv"

df = pd.read_csv(file_path)

print("=" * 60)
print("DATASET LOADED SUCCESSFULLY")
print("=" * 60)

# ------------------------------------------------------------
# 2. Display first 5 records
# ------------------------------------------------------------

print("\nFIRST 5 RECORDS")
print("=" * 60)
display(df.head())

# ------------------------------------------------------------
# 3. Display last 5 records
# ------------------------------------------------------------

print("\nLAST 5 RECORDS")
print("=" * 60)
display(df.tail())

# ------------------------------------------------------------
# 4. Display dataset dimensions
# ------------------------------------------------------------

rows, columns = df.shape

print("\nDATASET DIMENSIONS")
print("=" * 60)
print("Number of rows    :", rows)
print("Number of columns :", columns)

# ------------------------------------------------------------
# 5. Display column names
# ------------------------------------------------------------

print("\nCOLUMN NAMES")
print("=" * 60)

for i, column in enumerate(df.columns, start=1):
    print(f"{i}. {column}")

# ------------------------------------------------------------
# 6. Display data types
# ------------------------------------------------------------

print("\nDATA TYPES")
print("=" * 60)
print(df.dtypes)

# ------------------------------------------------------------
# 7. Display dataset information
# ------------------------------------------------------------

print("\nDATASET INFORMATION")
print("=" * 60)

df.info()

# ------------------------------------------------------------
# 8. Statistical summary of numerical variables
# ------------------------------------------------------------

print("\nSTATISTICAL SUMMARY")
print("=" * 60)
display(df.describe())

# ------------------------------------------------------------
# 9. Statistical summary including categorical variables
# ------------------------------------------------------------

print("\nCOMPLETE STATISTICAL SUMMARY")
print("=" * 60)
display(df.describe(include="all"))

# ------------------------------------------------------------
# 10. Display random sample
# ------------------------------------------------------------

print("\nRANDOM SAMPLE OF 10 RECORDS")
print("=" * 60)
display(df.sample(10, random_state=42))

# ------------------------------------------------------------
# 11. Check missing values
# ------------------------------------------------------------

print("\nMISSING VALUES")
print("=" * 60)

missing_values = df.isnull().sum()
display(missing_values)

# ------------------------------------------------------------
# 12. Check duplicate records
# ------------------------------------------------------------

print("\nDUPLICATE RECORDS")
print("=" * 60)

duplicate_count = df.duplicated().sum()

print("Number of duplicate records:", duplicate_count)

print("\n" + "=" * 60)
print("INITIAL DATASET INSPECTION COMPLETED")
print("=" * 60)
```

## Conclusion

The Indian Food dataset was successfully loaded into Python using the Pandas library.

The initial inspection included the dataset dimensions, column names, data types, first and last records, statistical summary, random sample, missing values, and duplicate records.

This initial analysis provides an understanding of the structure and quality of the dataset and prepares it for the next stages of the EDA process, including basic statistical analysis, missing-data handling, data cleaning, data transformation, and visualization.
