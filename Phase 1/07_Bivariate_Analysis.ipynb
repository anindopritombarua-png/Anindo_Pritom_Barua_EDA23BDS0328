### Objective

The objective of this notebook is to perform bivariate analysis on the Indian Food dataset. Bivariate analysis examines the relationship between two variables at a time.

Different visualization techniques such as scatter plots, boxplots, violin plots, grouped bar charts, and regression plots will be used to identify relationships, differences, and patterns between numerical and categorical variables.

```python
# ============================================================
# EDA PROJECT - BIVARIATE ANALYSIS
# Name        : ANINDO PRITOM BARUA
# Registration: 23BDS0328
# Dataset     : Indian Food Dataset
# ============================================================

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# ------------------------------------------------------------
# 1. Load the transformed dataset
# ------------------------------------------------------------

file_path = "indian_food_transformed.csv"

df = pd.read_csv(file_path)

print("=" * 70)
print("BIVARIATE ANALYSIS")
print("=" * 70)

print("\nDataset loaded successfully.")
print("Rows    :", df.shape[0])
print("Columns :", df.shape[1])

sns.set_theme(style="whitegrid")

# ------------------------------------------------------------
# 2. Visualization 1
# Preparation Time vs Cooking Time
# ------------------------------------------------------------

plt.figure(figsize=(10, 6))

sns.scatterplot(
    data=df,
    x="prep_time",
    y="cook_time"
)

plt.title("Preparation Time vs Cooking Time")
plt.xlabel("Preparation Time (minutes)")
plt.ylabel("Cooking Time (minutes)")
plt.tight_layout()
plt.show()

# ------------------------------------------------------------
# 3. Visualization 2
# Cooking Time by Diet Type
# ------------------------------------------------------------

plt.figure(figsize=(9, 6))

sns.boxplot(
    data=df,
    x="diet",
    y="cook_time"
)

plt.title("Cooking Time by Diet Type")
plt.xlabel("Diet Type")
plt.ylabel("Cooking Time (minutes)")
plt.xticks(rotation=20)
plt.tight_layout()
plt.show()

# ------------------------------------------------------------
# 4. Visualization 3
# Preparation Time by Course
# ------------------------------------------------------------

plt.figure(figsize=(10, 6))

sns.violinplot(
    data=df,
    x="course",
    y="prep_time"
)

plt.title("Preparation Time Distribution by Course")
plt.xlabel("Course")
plt.ylabel("Preparation Time (minutes)")
plt.xticks(rotation=20)
plt.tight_layout()
plt.show()

# ------------------------------------------------------------
# 5. Visualization 4
# Region vs Diet Type
# ------------------------------------------------------------
# Use the most common regions to keep the graph readable.

top_regions = (
    df["region"]
    .value_counts()
    .head(10)
    .index
)

region_diet = df[df["region"].isin(top_regions)]

plt.figure(figsize=(12, 6))

sns.countplot(
    data=region_diet,
    x="region",
    hue="diet"
)

plt.title("Diet Type Distribution Across Top 10 Regions")
plt.xlabel("Region")
plt.ylabel("Number of Recipes")
plt.xticks(rotation=45, ha="right")
plt.legend(title="Diet Type")
plt.tight_layout()
plt.show()

# ------------------------------------------------------------
# 6. Visualization 5
# Course vs Diet Type
# ------------------------------------------------------------

plt.figure(figsize=(10, 6))

sns.countplot(
    data=df,
    x="course",
    hue="diet"
)

plt.title("Diet Type Distribution Across Food Courses")
plt.xlabel("Course")
plt.ylabel("Number of Recipes")
plt.xticks(rotation=20)
plt.legend(title="Diet Type")
plt.tight_layout()
plt.show()

# ------------------------------------------------------------
# 7. Visualization 6
# Regression - Preparation Time vs Cooking Time
# ------------------------------------------------------------

plt.figure(figsize=(10, 6))

sns.regplot(
    data=df,
    x="prep_time",
    y="cook_time",
    scatter_kws={"alpha": 0.6},
    line_kws={}
)

plt.title("Regression Analysis: Preparation Time vs Cooking Time")
plt.xlabel("Preparation Time (minutes)")
plt.ylabel("Cooking Time (minutes)")
plt.tight_layout()
plt.show()

# ------------------------------------------------------------
# 8. Correlation between numerical variables
# ------------------------------------------------------------

print("\nCORRELATION BETWEEN NUMERICAL VARIABLES")
print("=" * 70)

correlation = df[
    ["prep_time", "cook_time", "total_time"]
].corr()

display(correlation)

# ------------------------------------------------------------
# 9. Interpretation of correlation
# ------------------------------------------------------------

print("\nCORRELATION INTERPRETATION")
print("=" * 70)

prep_cook_corr = correlation.loc["prep_time", "cook_time"]

print(
    f"Correlation between preparation time and cooking time: "
    f"{prep_cook_corr:.3f}"
)

if prep_cook_corr > 0:
    print("The relationship between preparation time and cooking time is positive.")
elif prep_cook_corr < 0:
    print("The relationship between preparation time and cooking time is negative.")
else:
    print("There is no linear relationship between preparation time and cooking time.")

print("\n" + "=" * 70)
print("BIVARIATE ANALYSIS COMPLETED")
print("=" * 70)
```

## Conclusion

Bivariate analysis was successfully performed on the Indian Food dataset to examine relationships between pairs of variables.

The relationship between preparation time and cooking time was examined using scatter and regression plots. Boxplots and violin plots were used to compare numerical variables across different diet and course categories.

Grouped bar charts were used to examine the relationship between categorical variables such as region, diet, and course.

The correlation analysis provided an additional numerical measure of the relationship between preparation time and cooking time. These analyses reveal relationships and patterns that cannot be observed by examining individual variables separately.

The dataset is now ready for multivariate analysis involving three or more variables simultaneously.
