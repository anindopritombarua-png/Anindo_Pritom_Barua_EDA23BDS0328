### Objective

The objective of this notebook is to clean and prepare the Indian Food dataset for further analysis. The cleaning process includes checking and removing duplicate records, removing unnecessary spaces from text values, standardizing text formats, handling inconsistent values, and verifying the structure and quality of the cleaned dataset.

```python
# ============================================================
# EDA PROJECT - DATA CLEANING
# Name        : ANINDO PRITOM BARUA
# Registration: 23BDS0328
# Dataset     : Indian Food Dataset
# ============================================================

import pandas as pd
import numpy as np

# ------------------------------------------------------------
# 1. Load the dataset after missing-value treatment
# ------------------------------------------------------------

file_path = "indian_food_missing_handled.csv"

df = pd.read_csv(file_path)

print("=" * 70)
print("DATA CLEANING")
print("=" * 70)

# ------------------------------------------------------------
# 2. Display original dataset information
# ------------------------------------------------------------

print("\nORIGINAL DATASET INFORMATION")
print("=" * 70)

print("Number of rows    :", df.shape[0])
print("Number of columns :", df.shape[1])

# ------------------------------------------------------------
# 3. Check duplicate records
# ------------------------------------------------------------

print("\nDUPLICATE RECORDS BEFORE CLEANING")
print("=" * 70)

duplicate_count = df.duplicated().sum()

print("Number of duplicate records:", duplicate_count)

# ------------------------------------------------------------
# 4. Remove duplicate records
# ------------------------------------------------------------

df = df.drop_duplicates().reset_index(drop=True)

print("\nDUPLICATE RECORDS AFTER CLEANING")
print("=" * 70)

print("Number of duplicate records:", df.duplicated().sum())

# ------------------------------------------------------------
# 5. Clean column names
# ------------------------------------------------------------

print("\nCLEANING COLUMN NAMES")
print("=" * 70)

# Remove leading/trailing spaces
df.columns = df.columns.str.strip()

# Convert column names to lowercase
df.columns = df.columns.str.lower()

# Replace spaces with underscores
df.columns = df.columns.str.replace(" ", "_", regex=False)

print("Cleaned column names:")
print(df.columns.tolist())

# ------------------------------------------------------------
# 6. Clean text columns
# ------------------------------------------------------------

print("\nCLEANING TEXT VALUES")
print("=" * 70)

categorical_columns = df.select_dtypes(include="object").columns.tolist()

for column in categorical_columns:

    # Remove leading and trailing spaces
    df[column] = df[column].astype(str).str.strip()

    # Replace multiple spaces with a single space
    df[column] = df[column].str.replace(
        r"\s+",
        " ",
        regex=True
    )

# ------------------------------------------------------------
# 7. Standardize selected categorical variables
# ------------------------------------------------------------

# Convert selected categorical variables to lowercase
# This prevents different capitalization from being treated
# as different categories.

columns_to_lowercase = [
    "diet",
    "flavor_profile",
    "course",
    "state",
    "region"
]

for column in columns_to_lowercase:
    if column in df.columns:
        df[column] = df[column].str.lower()

# ------------------------------------------------------------
# 8. Standardize food names
# ------------------------------------------------------------

if "name" in df.columns:
    df["name"] = df["name"].str.title()

# ------------------------------------------------------------
# 9. Convert numerical columns to numeric data types
# ------------------------------------------------------------

numerical_columns = [
    "prep_time",
    "cook_time",
    "total_time"
]

for column in numerical_columns:
    if column in df.columns:
        df[column] = pd.to_numeric(
            df[column],
            errors="coerce"
        )

# ------------------------------------------------------------
# 10. Check data types after cleaning
# ------------------------------------------------------------

print("\nDATA TYPES AFTER CLEANING")
print("=" * 70)

print(df.dtypes)

# ------------------------------------------------------------
# 11. Check missing values after cleaning
# ------------------------------------------------------------

print("\nMISSING VALUES AFTER CLEANING")
print("=" * 70)

display(df.isnull().sum())

# ------------------------------------------------------------
# 12. Check duplicates after cleaning
# ------------------------------------------------------------

print("\nDUPLICATES AFTER CLEANING")
print("=" * 70)

print("Number of duplicate records:", df.duplicated().sum())

# ------------------------------------------------------------
# 13. Display cleaned dataset
# ------------------------------------------------------------

print("\nCLEANED DATASET - FIRST 10 RECORDS")
print("=" * 70)

display(df.head(10))

# ------------------------------------------------------------
# 14. Display cleaned dataset information
# ------------------------------------------------------------

print("\nCLEANED DATASET INFORMATION")
print("=" * 70)

df.info()

# ------------------------------------------------------------
# 15. Save cleaned dataset
# ------------------------------------------------------------

output_file = "indian_food_cleaned.csv"

df.to_csv(output_file, index=False)

print("\n" + "=" * 70)
print("CLEANED DATASET SAVED SUCCESSFULLY")
print("=" * 70)

print("File name:", output_file)
print("Rows:", df.shape[0])
print("Columns:", df.shape[1])
```

The cleaning process included checking and removing duplicate records, standardizing column names, removing unnecessary spaces, standardizing categorical text values, formatting food names, and ensuring that numerical variables were stored using appropriate numeric data types.

The cleaned dataset was verified for remaining missing values and duplicate records and was saved as `indian_food_cleaned.csv`.

The resulting dataset is now ready for data transformation and exploratory visualization.
