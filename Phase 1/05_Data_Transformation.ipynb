### Objective

The objective of this notebook is to transform the cleaned Indian Food dataset into a form suitable for further analysis. New meaningful features will be created, numerical variables will be normalized and standardized, and logarithmic transformations will be applied where appropriate. These transformations help improve data comparability and provide additional information for exploratory analysis.

```python
# ============================================================
# EDA PROJECT - DATA TRANSFORMATION
# Name        : ANINDO PRITOM BARUA
# Registration: 23BDS0328
# Dataset     : Indian Food Dataset
# ============================================================

import pandas as pd
import numpy as np

# ------------------------------------------------------------
# 1. Load the cleaned dataset
# ------------------------------------------------------------

file_path = "indian_food_cleaned.csv"

df = pd.read_csv(file_path)

print("=" * 70)
print("DATA TRANSFORMATION")
print("=" * 70)

print("\nOriginal Dataset Shape:")
print("Rows    :", df.shape[0])
print("Columns :", df.shape[1])

# ------------------------------------------------------------
# 2. Create Total Time Feature
# ------------------------------------------------------------
# Total preparation + cooking time

if "prep_time" in df.columns and "cook_time" in df.columns:

    df["total_time"] = df["prep_time"] + df["cook_time"]

    print("\nTOTAL TIME FEATURE CREATED")
    print("=" * 70)

    display(
        df[["name", "prep_time", "cook_time", "total_time"]].head(10)
    )

# ------------------------------------------------------------
# 3. Create Quick Recipe Feature
# ------------------------------------------------------------
# A recipe taking 30 minutes or less is classified as quick.

if "total_time" in df.columns:

    df["quick_recipe"] = np.where(
        df["total_time"] <= 30,
        "Yes",
        "No"
    )

    print("\nQUICK RECIPE FEATURE")
    print("=" * 70)

    print(df["quick_recipe"].value_counts())

# ------------------------------------------------------------
# 4. Min-Max Normalization
# ------------------------------------------------------------
# Formula:
# (x - minimum) / (maximum - minimum)

numeric_columns = [
    "prep_time",
    "cook_time",
    "total_time"
]

for column in numeric_columns:

    if column in df.columns:

        minimum = df[column].min()
        maximum = df[column].max()

        if maximum != minimum:
            df[column + "_normalized"] = (
                (df[column] - minimum) /
                (maximum - minimum)
            )

# ------------------------------------------------------------
# 5. Z-Score Standardization
# ------------------------------------------------------------
# Formula:
# (x - mean) / standard deviation

for column in numeric_columns:

    if column in df.columns:

        mean_value = df[column].mean()
        std_value = df[column].std()

        if std_value != 0:

            df[column + "_standardized"] = (
                (df[column] - mean_value) /
                std_value
            )

# ------------------------------------------------------------
# 6. Logarithmic Transformation
# ------------------------------------------------------------
# log1p(x) is used because it safely handles zero values.

for column in numeric_columns:

    if column in df.columns:

        df[column + "_log"] = np.log1p(
            df[column].clip(lower=0)
        )

# ------------------------------------------------------------
# 7. Display transformed numerical variables
# ------------------------------------------------------------

print("\nTRANSFORMED NUMERICAL VARIABLES")
print("=" * 70)

transformed_columns = []

for column in df.columns:

    if (
        "_normalized" in column
        or "_standardized" in column
        or "_log" in column
    ):
        transformed_columns.append(column)

display(
    df[transformed_columns].head(10)
)

# ------------------------------------------------------------
# 8. Check normalized values
# ------------------------------------------------------------

print("\nNORMALIZED VALUE RANGE")
print("=" * 70)

for column in numeric_columns:

    normalized_column = column + "_normalized"

    if normalized_column in df.columns:

        print(
            f"{normalized_column}: "
            f"Minimum = {df[normalized_column].min():.2f}, "
            f"Maximum = {df[normalized_column].max():.2f}"
        )

# ------------------------------------------------------------
# 9. Check standardized values
# ------------------------------------------------------------

print("\nSTANDARDIZED VALUE SUMMARY")
print("=" * 70)

standardized_columns = [
    column + "_standardized"
    for column in numeric_columns
    if column + "_standardized" in df.columns
]

display(df[standardized_columns].describe())

# ------------------------------------------------------------
# 10. Display final transformed dataset
# ------------------------------------------------------------

print("\nFINAL TRANSFORMED DATASET")
print("=" * 70)

display(df.head(10))

print("\nFinal Dataset Shape:")
print("Rows    :", df.shape[0])
print("Columns :", df.shape[1])

# ------------------------------------------------------------
# 11. Save transformed dataset
# ------------------------------------------------------------

output_file = "indian_food_transformed.csv"

df.to_csv(output_file, index=False)

print("\n" + "=" * 70)
print("TRANSFORMED DATASET SAVED SUCCESSFULLY")
print("=" * 70)

print("File name:", output_file)
```

## Conclusion

The cleaned Indian Food dataset was successfully transformed using several data-transformation techniques.

A new `total_time` feature was created by combining preparation and cooking time. A `quick_recipe` categorical feature was also created to identify recipes that can be prepared within 30 minutes.

Min-Max normalization, Z-score standardization, and logarithmic transformation were applied to the numerical variables. These transformations improve the comparability of numerical variables and provide additional features for subsequent exploratory analysis.

The transformed dataset was saved as `indian_food_transformed.csv` and is ready for the visualization stages of the EDA process.
