### Objective

The objective of this notebook is to perform univariate analysis on the Indian Food dataset. Univariate analysis examines one variable at a time to understand its distribution, frequency, central tendency, and spread.

Both numerical and categorical variables will be analyzed using appropriate graphical techniques. Histograms, boxplots, and bar charts will be used to visualize the characteristics of individual variables.

```python
# ============================================================
# EDA PROJECT - UNIVARIATE ANALYSIS
# Name        : ANINDO PRITOM BARUA
# Registration: 23BDS0328
# Dataset     : Indian Food Dataset
# ============================================================

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# ------------------------------------------------------------
# 1. Load the transformed dataset
# ------------------------------------------------------------

file_path = "indian_food_transformed.csv"

df = pd.read_csv(file_path)

print("=" * 70)
print("UNIVARIATE ANALYSIS")
print("=" * 70)

print("\nDataset loaded successfully.")
print("Rows    :", df.shape[0])
print("Columns :", df.shape[1])

# Set visual style
sns.set_theme(style="whitegrid")

# ------------------------------------------------------------
# 2. Visualization 1 - Preparation Time Histogram
# ------------------------------------------------------------

plt.figure(figsize=(10, 6))

sns.histplot(
    data=df,
    x="prep_time",
    bins=20,
    kde=True
)

plt.title("Distribution of Preparation Time")
plt.xlabel("Preparation Time (minutes)")
plt.ylabel("Frequency")
plt.tight_layout()
plt.show()

# ------------------------------------------------------------
# 3. Visualization 2 - Cooking Time Histogram
# ------------------------------------------------------------

plt.figure(figsize=(10, 6))

sns.histplot(
    data=df,
    x="cook_time",
    bins=20,
    kde=True
)

plt.title("Distribution of Cooking Time")
plt.xlabel("Cooking Time (minutes)")
plt.ylabel("Frequency")
plt.tight_layout()
plt.show()

# ------------------------------------------------------------
# 4. Visualization 3 - Preparation Time Boxplot
# ------------------------------------------------------------

plt.figure(figsize=(10, 5))

sns.boxplot(
    x=df["prep_time"]
)

plt.title("Boxplot of Preparation Time")
plt.xlabel("Preparation Time (minutes)")
plt.tight_layout()
plt.show()

# ------------------------------------------------------------
# 5. Visualization 4 - Diet Distribution
# ------------------------------------------------------------

plt.figure(figsize=(8, 5))

diet_counts = df["diet"].value_counts()

sns.barplot(
    x=diet_counts.index,
    y=diet_counts.values
)

plt.title("Distribution of Diet Types")
plt.xlabel("Diet Type")
plt.ylabel("Number of Recipes")
plt.xticks(rotation=20)
plt.tight_layout()
plt.show()

# ------------------------------------------------------------
# 6. Visualization 5 - Course Distribution
# ------------------------------------------------------------

plt.figure(figsize=(10, 6))

course_counts = df["course"].value_counts()

sns.barplot(
    x=course_counts.index,
    y=course_counts.values
)

plt.title("Distribution of Food Courses")
plt.xlabel("Course")
plt.ylabel("Number of Recipes")
plt.xticks(rotation=20)
plt.tight_layout()
plt.show()

# ------------------------------------------------------------
# 7. Visualization 6 - Region Distribution
# ------------------------------------------------------------

plt.figure(figsize=(12, 6))

region_counts = df["region"].value_counts().head(15)

sns.barplot(
    x=region_counts.index,
    y=region_counts.values
)

plt.title("Top 15 Regions by Number of Recipes")
plt.xlabel("Region")
plt.ylabel("Number of Recipes")
plt.xticks(rotation=45, ha="right")
plt.tight_layout()
plt.show()

# ------------------------------------------------------------
# 8. Numerical Summary
# ------------------------------------------------------------

print("\nNUMERICAL VARIABLE SUMMARY")
print("=" * 70)

display(
    df[
        ["prep_time", "cook_time", "total_time"]
    ].describe()
)

# ------------------------------------------------------------
# 9. Categorical Frequency Summary
# ------------------------------------------------------------

print("\nCATEGORICAL VARIABLE FREQUENCIES")
print("=" * 70)

print("\nDiet:")
print(df["diet"].value_counts())

print("\nCourse:")
print(df["course"].value_counts())

print("\nRegion - Top 15:")
print(df["region"].value_counts().head(15))

print("\n" + "=" * 70)
print("UNIVARIATE ANALYSIS COMPLETED")
print("=" * 70)
```

## Conclusion

Univariate analysis was successfully performed on the Indian Food dataset.

The distributions of preparation time and cooking time were examined using histograms, while a boxplot was used to understand the spread and possible outliers in preparation time.

Categorical variables such as diet, course, and region were analyzed using bar charts to understand their frequency distributions.

The visualizations provide an initial understanding of individual variables and their distributions. These findings will be used as a foundation for examining relationships between multiple variables in the subsequent bivariate and multivariate analyses.
