### Objective

The objective of this notebook is to identify, analyze, and handle missing values present in the Indian Food dataset. The percentage of missing values in each column will be calculated, appropriate methods will be used to handle missing values, and the dataset will be verified after treatment.

```python
# ============================================================
# EDA PROJECT - HANDLING MISSING DATA
# Name        : ANINDO PRITOM BARUA
# Registration: 23BDS0328
# Dataset     : Indian Food Dataset
# ============================================================

import pandas as pd
import numpy as np

# ------------------------------------------------------------
# 1. Load the dataset
# ------------------------------------------------------------

file_path = "indian_food_23BDS0328_cleaned.csv"

df = pd.read_csv(file_path)

print("=" * 70)
print("HANDLING MISSING DATA")
print("=" * 70)

# ------------------------------------------------------------
# 2. Check missing values before treatment
# ------------------------------------------------------------

print("\nMISSING VALUES BEFORE TREATMENT")
print("=" * 70)

missing_before = df.isnull().sum()

display(missing_before.to_frame(name="Missing Values"))

# ------------------------------------------------------------
# 3. Calculate percentage of missing values
# ------------------------------------------------------------

print("\nPERCENTAGE OF MISSING VALUES")
print("=" * 70)

missing_percentage = (df.isnull().sum() / len(df)) * 100

missing_summary = pd.DataFrame({
    "Missing Values": df.isnull().sum(),
    "Missing Percentage": missing_percentage.round(2)
})

display(missing_summary)

# ------------------------------------------------------------
# 4. Identify numerical and categorical columns
# ------------------------------------------------------------

numerical_columns = df.select_dtypes(include=np.number).columns.tolist()
categorical_columns = df.select_dtypes(include="object").columns.tolist()

print("\nNUMERICAL COLUMNS")
print("=" * 70)
print(numerical_columns)

print("\nCATEGORICAL COLUMNS")
print("=" * 70)
print(categorical_columns)

# ------------------------------------------------------------
# 5. Handle missing numerical values
# ------------------------------------------------------------
# Numerical missing values are replaced using the median.
# Median is used because it is less affected by extreme values.

for column in numerical_columns:
    if df[column].isnull().sum() > 0:
        median_value = df[column].median()
        df[column] = df[column].fillna(median_value)

# ------------------------------------------------------------
# 6. Handle missing categorical values
# ------------------------------------------------------------
# Missing categorical values are replaced with "Unknown".

for column in categorical_columns:
    if df[column].isnull().sum() > 0:
        df[column] = df[column].fillna("Unknown")

# ------------------------------------------------------------
# 7. Check missing values after treatment
# ------------------------------------------------------------

print("\nMISSING VALUES AFTER TREATMENT")
print("=" * 70)

missing_after = df.isnull().sum()

display(missing_after.to_frame(name="Remaining Missing Values"))

# ------------------------------------------------------------
# 8. Compare before and after treatment
# ------------------------------------------------------------

print("\nBEFORE AND AFTER COMPARISON")
print("=" * 70)

comparison = pd.DataFrame({
    "Before Treatment": missing_before,
    "After Treatment": missing_after
})

display(comparison)

# ------------------------------------------------------------
# 9. Verify total missing values
# ------------------------------------------------------------

total_missing = df.isnull().sum().sum()

print("\nTOTAL REMAINING MISSING VALUES:", total_missing)

# ------------------------------------------------------------
# 10. Display sample of treated dataset
# ------------------------------------------------------------

print("\nSAMPLE OF DATASET AFTER MISSING VALUE TREATMENT")
print("=" * 70)

display(df.head(10))

# ------------------------------------------------------------
# 11. Save the dataset after missing-value treatment
# ------------------------------------------------------------

output_file = "indian_food_missing_handled.csv"

df.to_csv(output_file, index=False)

print("\nDataset saved successfully as:")
print(output_file)

print("\n" + "=" * 70)
print("MISSING DATA HANDLING COMPLETED")
print("=" * 70)
```

## Conclusion

The missing values present in the Indian Food dataset were successfully identified and analyzed.

The percentage of missing values was calculated for each column. Numerical missing values were handled using the median, while missing categorical values were replaced with the value "Unknown".

The dataset was then rechecked to verify that the missing values had been successfully handled. The processed dataset was saved as `indian_food_missing_handled.csv` for use in subsequent EDA tasks.

The dataset is now ready for the data-cleaning stage.
